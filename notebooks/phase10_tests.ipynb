## Test 1: RecursiveComputeGraph (RCG)

```python
# Test RCG - Graph Creation
graph = RecursiveComputeGraph("test_graph")
assert graph.name == "test_graph"
assert len(graph.nodes) == 0
print("✅ test_create_graph passed")

# Test RCG - Add Node
node = ComputeNode(node_id="n1", initial_state={"value": 1.0})
graph.add_node(node)
assert "n1" in graph.nodes
print("✅ test_add_node passed")

# Test RCG - Connect Nodes
n2 = ComputeNode(node_id="n2", initial_state={"value": 2.0})
graph.add_node(n2)
graph.connect("n1", "n2")
assert "n2" in graph.edges["n1"]
assert "n1" in graph.edges["n2"]
print("✅ test_connect_nodes passed")

# Test RCG - Compute Coherence
coherence = graph.compute_coherence("n1")
assert 0.0 <= coherence <= 1.0
print(f"✅ test_compute_coherence passed (coherence={coherence:.3f})")

# Test RCG - Stabilize Node
result = graph.stabilize_node("n1")
assert result.node_id == "n1"
assert result.iterations >= 0
print(f"✅ test_stabilize_node passed (iterations={result.iterations})")

# Test RCG - Compute Cycle
graph2 = RecursiveComputeGraph()
for i in range(3):
    node = ComputeNode(node_id=f"n{i}", initial_state={"value": float(i)})
    graph2.add_node(node)
graph2.connect("n0", "n1")
graph2.connect("n1", "n2")
results = graph2.compute()
assert len(results) == 3
print("✅ test_compute_cycle passed")

print("\n✅ All RCG tests passed (6/6)")
```

## Test 2: PositionalReferenceSystem (PRS)

```python
# Test PRS - Create System
prs = PositionalReferenceSystem("test_prs")
assert prs.name == "test_prs"
print("✅ test_create_system passed")

# Test PRS - Create Frame
frame = prs.create_frame("frame1")
assert frame.frame_id == "frame1"
print("✅ test_create_frame passed")

# Test PRS - Add Position
pos = Position(position_id="p1", coordinates={"x": 0.0, "y": 0.0})
frame.add_position(pos)
assert "p1" in frame.positions
print("✅ test_add_position passed")

# Test PRS - Transition
p2 = Position(position_id="p2", coordinates={"x": 1.0, "y": 1.0})
frame.add_position(p2)
result = frame.transition("p1", "p2")
assert result["valid"] is True
assert result["delta"]["x"] == 1.0
print("✅ test_transition passed")

# Test PRS - Compute Transition Path
for i in range(3):
    pos = Position(position_id=f"p{i}", coordinates={"x": float(i)})
    frame.add_position(pos)
path = prs.compute_transition_path("p0", "p2")
assert len(path) >= 1
print(f"✅ test_compute_transition_path passed (path length={len(path)})")

print("\n✅ All PRS tests passed (5/5)")
```

## Test 3: ResonancePropagationEngine (RPE)

```python
# Test RPE - Create Engine
rpe = ResonancePropagationEngine("test_rpe")
assert rpe.name == "test_rpe"
print("✅ test_create_engine passed")

# Test RPE - Register Node
rpe.register_node("n1", {"value": 1.0})
assert "n1" in rpe.nodes
print("✅ test_register_node passed")

# Test RPE - Connect and Propagate
rpe.register_node("n2", {"value": 1.0})
rpe.connect("n1", "n2")
rpe.set_constraint("n1", "value", 1.0)
result = rpe.propagate("n1")
assert result.source_id == "n1"
print("✅ test_connect_and_propagate passed")

# Test RPE - Field Coherence
coherence = rpe.compute_coherence("n1")
assert coherence == 1.0
print(f"✅ test_get_field_coherence passed (coherence={coherence})")

print("\n✅ All RPE tests passed (4/4)")
```

## Test 4: DynamicConstraintTopology (DCT)

```python
# Test DCT - Create Topology
dct = DynamicConstraintTopology("test_dct")
assert dct.name == "test_dct"
print("✅ test_create_topology passed")

# Test DCT - Add Node and Edge
dct.add_node("n1")
dct.add_node("n2")
dct.add_edge("n1", "n2")
assert "n1" in dct.nodes
assert ("n1", "n2") in dct.edges
print("✅ test_add_node_and_edge passed")

# Test DCT - Rewire
dct.set_coherence("n1", 0.9)
dct.set_coherence("n2", 0.9)
changes = dct.rewire()
assert len(changes) >= 0
print(f"✅ test_rewire passed (changes={len(changes)})")

# Test DCT - Get Metrics
metrics = dct.get_topology_metrics()
assert metrics["num_nodes"] == 2
assert metrics["num_edges"] == 1
print(f"✅ test_get_metrics passed (metrics={metrics})")

print("\n✅ All DCT tests passed (4/4)")
```
